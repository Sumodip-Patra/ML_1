```python
#as the data of journey is object i.e string we need to do feature engineering in the date of journey. Therefore we are splitting the data into day, month and year
final_df3['Day']= final_df3['Date_of_Journey'].str.split('/').str[0]
final_df3['Month']= final_df3['Date_of_Journey'].str.split('/').str[1]
final_df3['Year']= final_df3['Date_of_Journey'].str.split('/').str[2]
```

```python
final_df3
```

```python
#the same can be done using lambda function
final_df3['Day']= final_df3['Date_of_Journey'].apply(lambda x:x.split('/')[0])
final_df3['Month']= final_df3['Date_of_Journey'].apply(lambda x:x.split('/')[1])
final_df3['Year']= final_df3['Date_of_Journey'].apply(lambda x:x.split('/')[2])
```

```python
final_df3.head()
```

```python
#converting the day, month and year into integer
final_df3['Day']=final_df3['Day'].astype(int)
final_df3['Month']=final_df3['Month'].astype(int)
final_df3['Year']=final_df3['Year'].astype(int)
```

```python
final_df3.info()
```

```python
final_df3.drop('Date_of_Journey',axis=1,inplace=True)
```

```python
final_df3.head()
```

```python
#we do not need the date of the arrival only time is needed
final_df3['Arrival_Time']= final_df3['Arrival_Time'].apply(lambda x:x.split(' ')[0])
final_df3.head()
```

```python
#now dividing the arrival time into hours and minutes and then converting them into int
final_df3['Arrival_hour']= final_df3['Arrival_Time'].apply(lambda x:x.split(':')[0]).astype(int)
final_df3['Arrival_min']= final_df3['Arrival_Time'].apply(lambda x:x.split(':')[1]).astype(int)
```

```python
final_df3.drop('Arrival_Time',axis=1,inplace=True)
final_df3.head()
```

```python
final_df3.info()
```

```python
#doing the same with departure time
final_df3['Dep_hour']= final_df3['Dep_Time'].apply(lambda x:x.split(':')[0]).astype(int)
final_df3['Dep_min']= final_df3['Dep_Time'].apply(lambda x:x.split(':')[1]).astype(int)
```

```python
final_df3.drop('Dep_Time',axis=1,inplace=True)
final_df3.head()
```

```python
#Total_stops
final_df3['Total_Stops'].unique()
```

```python
from typing_extensions import final
final_df3['Total_Stops']= final_df3['Total_Stops'].map({'non-stop':0,'1 stop':1,'2 stops':2,'3 stops':3,'4 stops':4,'nan':1})
```

```python
final_df3.head()
```

```python
final_df3.drop('Route',axis=1,inplace=True)
final_df3.head(2)
```
