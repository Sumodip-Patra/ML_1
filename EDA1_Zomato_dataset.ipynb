```python
df_country= pd.read_excel('Country-Code.xlsx')
df_country.head()
```

```python
df_country.columns
```

```python
df_country.describe()
```

```python
#merging
final_df=pd.merge(df,df_country,on='Country Code',how='left')
final_df.head(2)
```

```python
final_df.types
```

```python
final_df.Country.value_counts()
```

```python
#array of the country name
country_name= final_df.Country.value_counts().index
country_name
```

```python
#array of the counts (transactions)
country_values= final_df.Country.value_counts().values
country_values
```

```python
#ploting the info in pie chart
plt.pie(country_values,labels=country_name,autopct='%1.2f')
```

```python
#plotting only top 3 countries
plt.pie(country_values[:3],labels=country_name[:3],autopct='%1.2f%%')
```

```python
ratings= final_df.groupby(['Aggregate rating', 'Rating color', 'Rating text']).size().reset_index().rename(columns={0:'Rating Count'})
```

```python
ratings.head()
```

```python
sns.barplot(x='Aggregate rating',y= 'Rating Count',data= ratings)
```

```python
import matplotlib
matplotlib.rcParams['figure.figsize']=(14,7)
sns.barplot(x='Aggregate rating',y= 'Rating Count',hue='Rating color',data= ratings,palette=['pink','red','orange','yellow','green','green'])
#
```

```python
##count plot
sns.countplot(x='Rating color', data= ratings,palette=['pink','red','orange','yellow','green','green'] )
```

the above graph shows that how many times the color appears in the ratings data, like in the data white appeared only once therefore it has count as 1.
It has nothing to do with the ratings count
