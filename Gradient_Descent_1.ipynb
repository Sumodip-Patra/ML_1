```python
from sklearn.datasets import make_regression
import numpy as np
```

```python
X,y= make_regression(n_samples=4, n_features=1,n_informative=1,n_targets=1,noise=80,random_state=13)
```

```python
import matplotlib.pyplot as plt
%matplotlib inline
plt.scatter(X,y)
```

```python
#applying ols
from sklearn.linear_model import LinearRegression
lr= LinearRegression()
lr.fit(X,y)
```

```python
lr.coef_
```

```python
lr.intercept_
```

```python
plt.scatter(X,y)
plt.plot(X,lr.predict(X),color='orange')
```

```python
# we need to find the intercept by using Gradient Descent
#assuming slope is constant m= 78.35
#assuming starting value of intercept b = 0
```

```python

y_pred= (m*X + 0).reshape(4)
```

```python
#plotting at b = 0
plt.scatter(X,y)
plt.plot(X,lr.predict(X),color='orange',label='OLS')
plt.plot(X,y_pred,color='#A3E3D7',label= 'b=0')
plt.legend()
```

```python
#first step
m = 78.35
b=0

loss_slope = -2*np.sum(y - m*X.ravel()-b)
loss_slope
```

```python
learnrate=0.1
step_size= loss_slope*learnrate
step_size
```

```python
#new intercept
b= b- step_size
b
```

```python
y_pred1= (m*X + b).reshape(4)
plt.scatter(X,y)
plt.plot(X,lr.predict(X),color='orange',label='OLS')
plt.plot(X,y_pred1,color= '#d99f2e',label='{}'.format(b))
plt.plot(X,y_pred,color='#A3E3D7',label= 'b=0')
plt.legend()
```

```python
#instead of find the gradient descent at individual step we will use epoch
m= 78.35
b= 4
learning_rate= 0.1
epochs =10

for i in range(epochs):
  loss_slope = -2*np.sum(y- m*X.ravel()-b)

  b= b- loss_slope*learning_rate

  y_pred = m*X + b

  plt.plot(X,y_pred)

plt.scatter(X,y)
```
